```python
import numpy as np

# Hàm mục tiêu
def func_grad(w):
    grad_w1 = 0.2 * w[0]
    grad_w2 = 4 * w[1]
    return np.array([grad_w1, grad_w2])

# Khởi tạo tham số
w = np.array([-5.0, -2.0],dtype = np.float32)  # Giá trị ban đầu của w1 và w2
v = np.array([0.0, 0.0])    # Giá trị ban đầu của momentum
alpha = 0.6                 # Learning rate
beta = 0.5                  # Momentum hệ số
epochs = 30

# Gradient Descent + Momentum
for epoch in range(epochs):
    grad = func_grad(w)
    v = beta * v + (1-beta) * grad
    w = w - alpha*v
    print(f"Epoch {epoch + 1}: w1 = {w[0]:.4f}, w2 = {w[1]:.4f}")

print("\nKết quả sau 30 epoch:")
print(f"w1 = {w[0]:.8f}, w2 = {w[1]:.10f}")
```

```python
import numpy as np

# Hàm mục tiêu và gradient
def func_grad(w):
    grad_w1 = 0.2 * w[0]
    grad_w2 = 4 * w[1]
    return np.array([grad_w1, grad_w2])

# Khởi tạo tham số
w = np.array([-5.0, -2.0])  # Giá trị ban đầu của w1 và w2
s = np.array([0.0, 0.0])    # Giá trị ban đầu của s1 và s2
alpha = 0.3                 # Learning rate
gamma = 0.9                 # Decay factor
epsilon = 1e-6              # Nhỏ tránh chia cho 0
epochs = 30

# RMSProp
for epoch in range(epochs):
    grad = func_grad(w)
    s = gamma * s + (1 - gamma) * grad**2
    w = w - alpha / np.sqrt(s + epsilon) * grad
    print(f"Epoch {epoch + 1}: w1 = {w[0]:.4f}, w2 = {w[1]:.4f}")

print("\nKết quả sau 30 epoch:")
print(f"w1 = {w[0]:.8f}, w2 = {w[1]:.20f}")
```

```python
import numpy as np

# Hàm mục tiêu và gradient
def func_grad(w):
    grad_w1 = 0.2 * w[0]
    grad_w2 = 4 * w[1]
    return np.array([grad_w1, grad_w2])

# Khởi tạo thông số
w = np.array([-5.0, -2.0])
v = np.array([0.0, 0.0])
s = np.array([0.0, 0.0])
alpha = 0.2
beta1 = 0.9
beta2 = 0.999
eps = 1e-6

# Adam optimization
for epoch in range(30):
    grad = func_grad(w)
    v = beta1 * v + (1 - beta1) * grad
    s = beta2 * s + (1 - beta2) * (grad ** 2)
    v_corr = v / (1 - beta1 ** (epoch + 1))
    s_corr = s / (1 - beta2 ** (epoch + 1))
    w -= alpha * v_corr / (np.sqrt(s_corr) + eps)
    print(f"Epoch {epoch + 1}: w1 = {w[0]:.4f}, w2 = {w[1]:.4f}")

print(f"w after 30 epochs: {w}")
```
